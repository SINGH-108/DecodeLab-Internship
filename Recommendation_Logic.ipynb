```python
movies = {
    "The Dark Knight": ["Action", "Crime", "Drama"],
    "Interstellar": ["Sci-Fi", "Drama", "Adventure"],
    "Inception": ["Action", "Sci-Fi", "Heist"],
    "The Godfather": ["Crime", "Drama"],
    "Superbad": ["Comedy", "Teen"],
    "Finding Nemo": ["Animation", "Adventure", "Family"]
}
```

```python
vocabulary = sorted(list(set([tag for movie_tags in movies.values() for tag in movie_tags])))
print(f"Shared Vocabulary: {vocabulary}")
def vectorize(tags, vocab):
    return [1 if word in tags else 0 for word in vocab]
movie_vectors = {name: vectorize(tags, vocabulary) for name, tags in movies.items()}
```

```python
user_interests = ["Action", "Sci-Fi"]
user_vector = vectorize(user_interests, vocabulary)
print(f"User Vector: {user_vector}")
```

```python
def calculate_jaccard(vec1, vec2):
    intersection = sum([1 for i in range(len(vec1)) if vec1[i] == 1 and vec2[i] == 1])
    union = sum([1 for i in range(len(vec1)) if vec1[i] == 1 or vec2[i] == 1])
    return intersection / union if union != 0 else 0
scores = {name: calculate_jaccard(user_vector, vec) for name, vec in movie_vectors.items()}
```

```python
recommended_movies = sorted(scores.items(), key=lambda x: x[1], reverse=True)
print("Your Top-3 Recommendations")
for i in range(3):
    movie_name, score = recommended_movies[i]
    print(f"{i+1}. {movie_name} (Match Score: {round(score, 2)})")
```

```python
available_tags = ", ".join(vocabulary)
print(f"Available Genres: {available_tags}")
user_input = input("Enter genres you like (separated by commas): ").strip().split(",")
user_interests = [tag.strip().title() for tag in user_input]
user_vector = vectorize(user_interests, vocabulary)
```
